```python
import plotly.express as px
from plotnine.data import mpg
import plotly.graph_objects as go
import pandas as pd

fig1 = px.scatter()
fig1.update_traces() #fig1 es el scatter plot de puntos naranja con transparencia 0.2 y sin interaccción

gb = mpg.groupby(["cty", "hwy"])["manufacturer"].agg("count").reset_index()

fig2 = px.scatter()
fig2.update_traces() #fig2 es el scatter plot invisible con interacción mostrando el número de muestras que hay en cada punto (usa gb)



fig = go.Figure(data = fig1.data + fig2.data)
fig.show()
```
